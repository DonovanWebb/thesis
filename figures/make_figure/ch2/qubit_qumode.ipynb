```python
import numpy as np
import matplotlib.pyplot as plt

# Draw parabola with lines for quantum harmonic oscillator potential

x=np.linspace(-1, 1, 100)
y=x**2
spacing = 0.25
vlines = np.arange(spacing/2, 1.0, spacing)

# plot parabola but fade away for y getting larger
plt.figure(figsize=(6, 6))
plt.plot(x, y, color='black', alpha=0.5, linestyle='-', linewidth=7.5)
for i, v in enumerate(vlines):
    x_min = -np.sqrt(v)
    x_max = np.sqrt(v)
    alpha = 0.6 - 0.5 * (i / len(vlines))  # increase alpha with each line
    plt.plot([x_min, x_max], [v, v], color='black', linestyle='-', label=f'V = {v}', alpha=alpha, linewidth=4.5)
# plt dot dot dot above the last line
ofs = 0.07
plt.scatter([0,0,0], [vlines[-1] + ofs, vlines[-1] + 2*ofs, vlines[-1] + 3*ofs], color='black', s=60, alpha=0.5)
# remove axis ticks and labels
plt.xticks([])
plt.yticks([])
# remove axis spines
for spine in ['left', 'bottom', 'right', 'top']:
    plt.gca().spines[spine].set_visible(False)
f_name = "..\\..\\pdf_figure\\ch2\\qumode.pdf"
plt.savefig(f_name, bbox_inches='tight', pad_inches=0.1, transparent=True)
```

```python
# Jaynes-Cummings and anti-Jaynes-Cummings Hamiltonian

t = np.linspace(0, 2*np.pi, 100)
W = 1.0
S_JC = np.sin(W*t/2)**2 # spin up probability for Jaynes-Cummings model
N_JC = np.cos(W*t/2)**2 # N 

fig, axs = plt.subplots(2,2, figsize=(10, 8))

axJC = axs[:,0]
axAJC = axs[:,1]
axJC[0].plot(t, S_JC, label=r'Qubit', color='blue', linewidth=2)
axJC[1].plot(t, N_JC, label=r'Qumode', color='orange', linewidth=2)
axJC[0].set_xticks([0, np.pi/2, np.pi, 3*np.pi/2, 2*np.pi])
axJC[0].set_yticks([0, 1], 
               labels=[r'$P_\downarrow$', r'$P_\uparrow$'])
axJC[1].set_yticks([0, 1], 
               labels=[r'$P_n$', r'$P_{n+1}$'])
```
